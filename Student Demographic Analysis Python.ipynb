```python
import random as rn
import numpy as np
```

```python
rn.seed(12345)
ps21['Random'] = np.random.rand(200)
ps21.head()
ps21sort = ps21.sort_values(by = ['Random'])
print(ps21sort.head())
ps21sample = ps21sort[:60]
len(ps21sample)
```

```python
from scipy import stats as sp

var = ps21sample['Height']
c_lev = 0.90
units = "inches"

alpha = 1 - c_lev
n = len(var)
mean = round(var.mean(), 2)
sd = np.std(var, ddof = 1)
tstar = sp.t.ppf(1-(alpha/2), n-1)
me = round(tstar*sd/np.sqrt(n), 2)
lower = round(mean-me, 2)
upper = round(mean+me, 2)
print("In a sample of", n, "observations, the sample mean was", mean, units)
print("The margin of error was", me, units)
print("The", c_lev*100, "% confidence interval for the population mean")
print("is between", lower, "and", upper, units)
```
